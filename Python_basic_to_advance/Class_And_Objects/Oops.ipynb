## Classes and Objects
Object-Oriented Programming (OOP) is a programming paradigm that uses "objects" to design applications and computer programs. OOP allows for modeling real-world scenarios using classes and objects. This lesson covers the basics of creating classes and objects, including instance variables and methods.

```python
### A class is a blue print for creating objects. Attributes,methods
class Car:
    pass
    
audi=Car()   # object
bmw=Car()
print(type(audi))
```

```python
print(audi)
print(bmw)
```

```python
audi.windows=4 # window is a type of attribute that is present in the car
print(audi.windows)
```

```python
tata=Car()
tata.doors=4
print(tata.windows)
```

```python
### instance variables and methods
class Dog:
    ## constructor
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
 ## create objects 
dog1=Dog('Tommy',10)
dog2=Dog('Jack',11)
print(dog1)
print(dog1.name)
print(dog1.age)
print(dog2)
print(dog2.name)
print(dog2.age)
```

```python
##  Define class with instance methods
class Dog:
    def __init__(self,name,age):
        self.name=name #instance variable
        self.age=age
    def bark(self):
        print(f"{self.name} says Woof!!")
dog1 = Dog('Tommy',10)
dog1.bark()
dog2 = Dog('Jack',10)
dog2.bark()
        
```

```python
## Modeling a Bank account
## Define a class for bank account
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited new balance is {self.balance}")

    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient balance")
        else:
            self.balance-=amount
            print(f"{amount} is withdraw succesfully . Your  new balance is {self.balance}")
    def get_balance(self):
        return self.balance

# create an account
account=BankAccount('Mayuri',500000)
account.deposit(30000)
account.withdraw(1000000)
account.withdraw(10000)
account.get_balance()
```

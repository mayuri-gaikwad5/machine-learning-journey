# Decorators 
Decorators are a powerful and flexible feature in python that allows to modify the behaviour of a function or class method . they are commonly used to add functionality to functions or methods without modifying their actual code . this lesson covers the basics of decorators , including how to create and use them.

```python
## Function copy
## closures
## decorators
```

```python
## function copy 
def welcome():
    return "Welcome to the advanced python course"
welcome()
```

```python
wel = welcome
print(wel())
del welcome
print(wel())
```

```python
## closures -  functions inside a function
def main_welcome():
    msg = "welcome"
    def sub_welcome_method():
        print("welcome to the advance python course ")
        print(msg)
        print("Please learn these concepts properly")
    return sub_welcome_method()
main_welcome()
```

```python
## closures -  functions inside a function
def main_welcome(func):
    msg = "welcome"
    def sub_welcome_method():
        print("welcome to the advance python course ")
        func("welcome eveyone to this tutorial")
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
main_welcome(print)
```

```python
def main_welcome(func,lst):
    msg = "welcome"
    def sub_welcome_method():
        print("welcome to the advance python course ")
        print(func(lst))
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
main_welcome(len,[1,2,3,4,5,6])
```

```python
### Decorator 
def main_welcome(func):
    msg = "welcome"
    def sub_welcome_method():
        print("welcome to the advance python course ")
        func()
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
def course_introduction():
    print("This is an advance python course")
course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("This is an advance python course")
```

```python
## Decorator example 2
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called")
        func()
        print("Something is happening after the function is called")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
# Decorators with arguments 
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args , **kwargs)
        return wrapper
    return decorator 
```

```python
@repeat(3)
def say_hello():
    print("hello")
```

```python
say_hello()
```

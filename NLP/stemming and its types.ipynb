# Stemming 

stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or the root of words known as lemmma . Stemming is important in natural language understanding (NLU) and natural language processing(NLP)

```python
# Classification Problem 
## comments of products is a positive review or negative review

# Reviews-----> eating,eat,eaten[going,gone,goes]
# stem=eat,go 
```

# PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
words=['eating','eats','eaten','writing','writes','programming','programs','history','finally','finalized']
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
stemming.stem('sitting')
```

# RegexpStemmer Class"

NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms . It basically takes a single regular expression and removes any prefix or suffix that matches the expression . Let us seen an example

```python
from nltk.stem import RegexpStemmer
```

```python
reg_Stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_Stemmer.stem('eating')
```

```python
reg_Stemmer.stem('ingeating')
```

# snowball stemmer

```python
from nltk.stem import SnowballStemmer 
```

```python
snowballsstemmer=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"--->"+snowballsstemmer.stem(word))
```

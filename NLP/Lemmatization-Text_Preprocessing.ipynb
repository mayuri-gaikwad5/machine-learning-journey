# Wordnet Lemmatizer 

Lemmitization technique is like stemming. The output we will get after lemmitization is called 'lemma', which is a root word rather than root stem, the output of stemming . After lemmatization, we will be getting a valid word that means the same thing .

NLTK provides WordNetLemmatizer class which is a thin erapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma . Let us understand it with a example.



```python
## Q&A , Chatbots ,text summarization
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer =WordNetLemmatizer()
```

```python
''' :param pos: The Part Of Speech tag. Valid options are `"n"` for nouns,
    `"v"` for verbs, `"a"` for adjectives, `"r"` for adverbs and `"s"`
    for satellite adjectives.'''

lemmatizer.lemmatize("going",pos='v')
```

```python
words=['eating','eats','eaten','writing','writes','programming','programs','history','finally','finalized']
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
lemmatizer.lemmatize("fairly",pos='v'),lemmatizer.lemmatize("sportingly",pos='v')
```

 # Pandas-DataFrame And Series
- Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning.
- It provides two primary data structures: Series and DataFrame.
-  A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
## Series
''' A Pandas Series is a one-dimensional array-like object that can hold any data type. 
It is similar to a column in a table. '''

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
## Create a series from dictionary 
data = {'a' :1 ,'b':2 , 'c':3}
series_dic = pd.Series(data)
print(series_dic)
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data , index)
```

```python
## Dataframe 
## Create a DataFrame from a dictionary off list 

data = {
    'Name':['Mayuri' , 'Gouri' , 'Shrushti'],
    'Age':[21,19,25],
    'city':['Solapur','Pune','Nagpur']
}
df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a Data frame From a List of Dictionaries

data=[
    {'Name':'Mayuri','Age':20,'City':'Solapur'},
    {'Name':'SOham','Age':14,'City':'Delhi'},
    {'Name':'Pranav','Age':19,'City':'Pune'},
    {'Name':'Shrushti','Age':19,'City':'Nagapur'}
    
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```

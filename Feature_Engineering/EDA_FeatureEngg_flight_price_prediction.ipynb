```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Feature Engineering 
df['Date']=df['Date_of_Journey'].str.split('/').str[0]
```

```python
df['Month']=df['Date_of_Journey'].str.split('/').str[1]
```

```python
df['Year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df['Date']=df['Date'].astype(int)
```

```python
df['Month']=df['Month'].astype(int)
```

```python
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
df['Arrival_Time']
```

```python
df['Arrival_hour']=df['Arrival_Time'].str.split(':').str[0]
```

```python
df['Arrival_min']=df['Arrival_Time'].str.split(':').str[1]
```

```python
df['Arrival_min']
```

```python
df['Arrival_hour']=df['Arrival_hour'].astype(int)
df['Arrival_min']=df['Arrival_min'].astype(int)
```

```python
df.info()
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Departure_hour']=df['Dep_Time'].str.split(':').str[0]
df['Departure_min']=df['Dep_Time'].str.split(':').str[1]
```

```python
df.head()
```

```python
df['Departure_hour']=df['Departure_hour'].astype(int)
df['Departure_min']=df['Departure_min'].astype(int)
```

```python
df.info()
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops'].mode()
```

```python
df['Total_Stops'].isnull().sum()
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,'4 stops':4,np.nan:1})
```

```python
df['Total_Stops'].isnull().sum()
```

```python
df.head(2)
```

```python
df.drop('Route',axis=1,inplace=True)
```

```python
df.head(2)
```

```python
df['duration_hour']=df['Duration'].str.split(' ').str[0].str.split('h').str[0]
```

```python
df['duration_min']=df['Duration'].str.split(' ').str[1].str.split('m').str[0]
```

```python
df['duration_min'].isnull().sum()
```

```python
df['duration_min']=df['duration_min'].fillna(0)
```

```python
df['duration_min'].isnull().sum()
```

```python
df.info()
```

```python
df.drop('Duration',axis=1,inplace=True)
```

```python
# one hot encoding 
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder=OneHotEncoder()
```

```python
encoder.fit_transform(df[['Airline','Source','Additional_Info','Destination']]).toarray()
```

```python
encoder_df=pd.DataFrame(encoder.fit_transform(df[['Airline','Source','Additional_Info','Destination']]).toarray(),columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
df=pd.concat([df,encoder_df],axis=1)
```

```python
df.drop('Airline',axis=1,inplace=True)
df.drop('Source',axis=1,inplace=True)
df.drop('Additional_Info',axis=1,inplace=True)
df.drop('Destination',axis=1,inplace=True)
```

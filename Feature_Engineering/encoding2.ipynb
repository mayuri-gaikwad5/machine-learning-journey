# Ordinal Encoding

```python
# 1. high School :1
# 2. College :2
# 3. Graduate:3
# 4. Post-graduate:4
```

```python
## Ordinal Encoding
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({
    'size':['small','medium','large','medium','small','large']
})
```

```python
df
```

```python
## create an instance of OrdinalEncoder and the fit_transform 

encoder=OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
encoder.transform([['small']])
```

# Target Guided Ordinal Encoding 
we replace each category in the categorical variable with a numerical variable with a numerical value based 
on mean or median of the target variable for that category .


```python
import pandas as pd
# create a sample datframe with a categorical variable and a target variable

df=pd.DataFrame({
    'city':['New York','London','Paris','Tokyo','New York','Paris'],
    'price':[200,150,300,250,180,320]
})
```

```python
df
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded']=df['city'].map(mean_price)
```

```python
df
```

```python
df[['price','city_encoded']]
```
